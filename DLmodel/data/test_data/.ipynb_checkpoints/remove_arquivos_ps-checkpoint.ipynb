```python
import os

def remove_arquivos_ps(diretorio):
    try:
        # Listar todos os arquivos no diretório
        arquivos = os.listdir(diretorio)

        # Percorrer a lista de arquivos
        for arquivo in arquivos:
            # Verificar se o arquivo tem a extensão ".ps"
            if arquivo.endswith(".ps"):
                caminho_arquivo = os.path.join(diretorio, arquivo)
                # Remover o arquivo
                os.remove(caminho_arquivo)
                print(f"Arquivo {caminho_arquivo} removido com sucesso.")
    except Exception as e:
        print(f"Erro ao remover arquivos: {e}")

# Defina o diretório onde deseja remover os arquivos .ps
diretorio_alvo = "2_riboswitch"
remove_arquivos_ps(diretorio_alvo)
```
