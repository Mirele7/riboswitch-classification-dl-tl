```python
import random

def read_fasta(file_path):
    sequences = []
    current_sequence = ""

    with open(file_path, "r") as file:
        for line in file:
            if line.startswith(">"):
                if current_sequence:
                    sequences.append((header, current_sequence))
                    current_sequence = ""
                header = line.strip()
            else:
                current_sequence += line.strip()

        if current_sequence:
            sequences.append((header, current_sequence))

    return sequences

def fisher_yates_shuffle(sequence_list):
    for i in range(len(sequence_list) - 1, 0, -1):
        j = random.randint(0, i)
        sequence_list[i], sequence_list[j] = sequence_list[j], sequence_list[i]

    return sequence_list

def select_random_sequences(sequences, num_sequences):
    shuffled_sequences = fisher_yates_shuffle(sequences)
    selected_sequences = shuffled_sequences[:num_sequences]

    return selected_sequences

def write_fasta(output_file, selected_sequences):
    with open(output_file, "w") as file:
        for header, sequence in selected_sequences:
            file.write(f"{header}\n{sequence}\n")


# Substitua 'caminho/do/arquivo.fasta' pelo caminho do seu arquivo FASTA
file_path = 'train_genomic_background.fasta'
#file_path = 'train_sequences_riboswitch_ID.fasta'

# Substitua 100 pelo número desejado de sequências aleatórias a serem selecionadas
num_random_sequences = 400

# Lê as sequências do arquivo FASTA
all_sequences = read_fasta(file_path)

# Seleciona 100 sequências aleatórias
random_sequences = select_random_sequences(all_sequences, num_random_sequences)

# Substitua 'caminho/do/novo_arquivo.fasta' pelo caminho desejado para o novo arquivo FASTA
output_file_path = '400/train_400_genomic_background.fasta'
#output_file_path = '400/train_400_riboswitch.fasta'

# Escreve as sequências selecionadas no novo arquivo FASTA
write_fasta(output_file_path, random_sequences)

print(f"Sequências selecionadas foram salvas em '{output_file_path}'")

```
