```python
# Importar os módulos necessários
from Bio import SeqIO

def find_duplicate_sequences(file):
    """Encontra sequências duplicadas em um arquivo FASTA."""
    sequences = {}
    duplicates = set()
    
    for record in SeqIO.parse(file, "fasta"):
        seq = str(record.seq)
        if seq in sequences:
            duplicates.add(seq)
        else:
            sequences[seq] = record.id
            
    return duplicates

def compare_sequences(file1, file2):
    """Compara sequências de dois arquivos FASTA e encontra as iguais."""
    sequences_file1 = {str(record.seq): record.id for record in SeqIO.parse(file1, "fasta")}
    sequences_file2 = {str(record.seq): record.id for record in SeqIO.parse(file2, "fasta")}
    
    common_sequences = set(sequences_file1.keys()).intersection(sequences_file2.keys())
    
    return common_sequences

def main(file1, file2):
    print("Verificando sequências duplicadas no próprio arquivo...")

    duplicates_file1 = find_duplicate_sequences(file1)
    if duplicates_file1:
        print(f"Sequências duplicadas encontradas em {file1}:")
        for seq in duplicates_file1:
            print(seq)
    else:
        print(f"Nenhuma sequência duplicada encontrada em {file1}.")

    duplicates_file2 = find_duplicate_sequences(file2)
    if duplicates_file2:
        print(f"Sequências duplicadas encontradas em {file2}:")
        for seq in duplicates_file2:
            print(seq)
    else:
        print(f"Nenhuma sequência duplicada encontrada em {file2}.")

    print("\nComparando sequências entre os dois arquivos...")
    common_sequences = compare_sequences(file1, file2)
    if common_sequences:
        print("Sequências iguais encontradas entre os dois arquivos:")
        for seq in common_sequences:
            print(seq+'\n')
    else:
        print("Nenhuma sequência igual encontrada entre os dois arquivos.")

# Substitua pelos caminhos dos seus arquivos FASTA
file1 = "1000/val_1000.fasta"
file2 = "1000/train_1000.fasta"
main(file1, file2)

```

```python
from Bio import SeqIO

def read_fasta(file_path):
    sequences = {}
    with open(file_path, 'r') as file:
        for record in SeqIO.parse(file, 'fasta'):
            sequences[record.id] = str(record.seq)
    return sequences

def compare_sequences(file1_path, file2_path):
    sequences1 = read_fasta(file1_path)
    sequences2 = read_fasta(file2_path)

    for id1, seq1 in sequences1.items():
        for id2, seq2 in sequences2.items():
            if seq1 == seq2:
                print(f"Sequência id {id1} do arquivo 1 é igual à sequência id {id2} do arquivo 2.")

if __name__ == "__main__":
    file1_path = "100/train_100.fasta"
    file2_path = "100/val_100.fasta"

    compare_sequences(file1_path, file2_path)
```
