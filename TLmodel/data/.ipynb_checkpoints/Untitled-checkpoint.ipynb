```python
from Bio import SeqIO

# Lista de palavras para filtrar IDs
palavras_filtradas = ["CD-box", "HACA-box", "tRNA", "rRNA", "scaRNA"]

# Nome do arquivo de entrada FASTA
entrada_fasta = "train_1000.fasta"

# Nome do arquivo de saída FASTA
saida_fasta = "saida.fasta"

# Função para verificar se o ID termina com alguma das palavras filtradas
def id_termina_com_palavra(id, palavras):
    for palavra in palavras:
        if id.endswith(palavra):
            return True
    return False

# Lista para armazenar as sequências filtradas
sequencias_filtradas = []

# Abre o arquivo de entrada e lê as sequências
with open(entrada_fasta, "r") as fasta_file:
    for record in SeqIO.parse(fasta_file, "fasta"):
        if id_termina_com_palavra(record.id, palavras_filtradas):
            sequencias_filtradas.append(record)

# Escreve as sequências filtradas em um novo arquivo FASTA
with open(saida_fasta, "w") as output_file:
    SeqIO.write(sequencias_filtradas, output_file, "fasta")

print(f"Sequências com IDs terminando em {', '.join(palavras_filtradas)} foram salvas em {saida_fasta}")

```
