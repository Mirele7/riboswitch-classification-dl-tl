```python
###Remoneia o id das sequencias
def rename_sequence_header(header,name):
    # Substitui espaços em branco por "_"
    header = header.replace(" ", "_")
    # Adiciona "RNA" no final da string
    header = header + name
    return header

def read_fasta_file(file_path):
    sequences = {}
    current_seq_name = None
    with open(file_path, "r") as file:
        for line in file:
            line = line.strip()
            if not line:
                continue
            if line.startswith(">"):  # Header line
                current_seq_name = line[1:]
                sequences[current_seq_name] = ""
            else:  # Sequence line
                sequences[current_seq_name] += line
    return sequences

def write_fasta_file(file_path, sequences,name):
    with open(file_path, "w") as file:
        for seq_name, seq in sequences.items():
            header = f">{rename_sequence_header(seq_name,name)}\n"
            file.write(header)
            seq_lines = [seq[i:i+80] for i in range(0, len(seq), 80)]
            file.write("\n".join(seq_lines) + "\n")

if __name__ == "__main__":
    #lista_test =  ["tRNA_test.fasta","CD_test.fasta","HACA_test.fasta",
    #         "leader_test.fasta","5_8S_rRNA_test.fasta","5S_rRNA_test.fasta",
    #         "miRNA_test.fasta","riboswitch_test.fasta","ribozyme_test.fasta",
    #         "scaRNA_test.fasta"]
    #lista_name = [" tRNA"," CD-box"," HACA-box"," leader"," 5_8S_rRNA"," 5S_rRNA",
    #         " miRNA"," riboswitch"," ribozyme"," scaRNA"]
    #for i in range(10):
    input_file_path = "test_genomic_HIT.fasta"#+lista_test[i]  # Substitua pelo caminho do arquivo FASTA de entrada
    output_file_path = "test_genomic_background.fasta"#+lista_test[i]  # Substitua pelo caminho do arquivo FASTA de saída

    sequences = read_fasta_file(input_file_path, )
    write_fasta_file(output_file_path, sequences,' genomic_background')
```

```python
def read_fasta(file_path):
    sequences = {}
    current_header = None

    with open(file_path, 'r') as file:
        for line in file:
            line = line.strip()
            if line.startswith('>'):
                current_header = line[1:]
                sequences[current_header] = ''
            else:
                sequences[current_header] += line

    return sequences

def write_fasta(output_file, sequences):
    with open(output_file, 'w') as file:
        for header, sequence in sequences.items():
            file.write(f'>{header}\n{sequence}\n')

def merge_fasta_files(input_files, output_file):
    merged_sequences = {}

    for file_path in input_files:
        sequences = read_fasta(file_path)
        merged_sequences.update(sequences)

    write_fasta(output_file, merged_sequences)

# Lista de arquivos FASTA que deseja juntar
input_files = ['riboswitch.fasta','5S_rRNA.fasta','scaRNA.fasta','miRNA.fasta','IRES.fasta','ribozyme.fasta','CD.fasta','tRNA.fasta','HACA.fasta','leader.fasta','5_8S_rRNA.fasta']

# Nome do arquivo de saída
output_file = 'test_11_mirele.fasta'

# Juntar os arquivos
merge_fasta_files(input_files, output_file)

print(f'Arquivos FASTA foram mesclados com sucesso no arquivo {output_file}.')
```
