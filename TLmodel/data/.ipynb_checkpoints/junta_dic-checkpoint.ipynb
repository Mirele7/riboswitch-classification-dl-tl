```python
import pickle

def juntar_dicionarios(arquivo1, arquivo2, arquivo_saida):
    try:
        # Carrega os dicionários dos arquivos .pkl
        with open(arquivo1, 'rb') as f1:
            dict1 = pickle.load(f1)

        with open(arquivo2, 'rb') as f2:
            dict2 = pickle.load(f2)

        # Junte os dicionários, com as chaves do primeiro dicionário prevalecendo
        dict_resultante = {**dict2, **dict1}

        # Salve o dicionário resultante em um novo arquivo .pkl
        with open(arquivo_saida, 'wb') as f_saida:
            pickle.dump(dict_resultante, f_saida)

        print("Dicionários juntados com sucesso e o resultado foi salvo em", arquivo_saida)

    except Exception as e:
        print("Ocorreu um erro:", e)

# Exemplo de uso
arquivo1 = '100/foldings_100.pkl'
arquivo2 = 'foldings_ALL.pkl'
arquivo_saida = 'foldings_ALL_ALL.pkl'

juntar_dicionarios(arquivo1, arquivo2, arquivo_saida)
```

```python
def process_fasta(input_file, output_file):
    with open(input_file, 'r') as infile, open(output_file, 'w') as outfile:
        for line in infile:
            if line.startswith('>'):
                # Verifica se "riboswitch" já está presente no final da linha
                if not line.strip().endswith("riboswitch"):
                    line = line.rstrip() + " riboswitch\n"
            outfile.write(line)

if __name__ == "__main__":
    input_file = "train_sequences_riboswitch.fasta"  # Substitua pelo nome do seu arquivo fasta
    output_file = "train_sequences_riboswitch_ID.fasta"  # Substitua pelo nome do arquivo de saída desejado
    process_fasta(input_file, output_file)
```
