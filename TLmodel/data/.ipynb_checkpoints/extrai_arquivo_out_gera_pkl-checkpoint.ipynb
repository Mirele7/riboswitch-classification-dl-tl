```python
def format_fasta_single_line(input_file, output_file):
    with open(input_file, 'r') as infile, open(output_file, 'w') as outfile:
        sequence = ''
        header = None
        for line in infile:
            line = line.strip()
            if line.startswith('>'):
                if header is not None:
                    outfile.write(header + '\n')
                    outfile.write(sequence + '\n')
                header = line
                sequence = ''
            else:
                sequence += line
        # Write the last sequence
        if header is not None:
            outfile.write(header + '\n')
            outfile.write(sequence + '\n')

# Exemplo de uso:
#format_fasta_single_line('100/train_100.fa', '100/train_100_riboswitch.fa')
#format_fasta_single_line('100/test_100.fa', '100/test_100_riboswitch.fa')
```

```python
from collections import defaultdict

def encontrar_headers_duplicados(fasta_path):
    headers = defaultdict(int)

    with open(fasta_path, 'r') as f:
        for line in f:
            line = line.strip()
            if line.startswith('>'):
                headers[line] += 1

    duplicados = {h: count for h, count in headers.items() if count > 1}

    if duplicados:
        print("⚠️ Headers duplicados encontrados:\n")
        for h, count in duplicados.items():
            print(f"{h}  →  {count} vezes")
    else:
        print("✅ Nenhum header duplicado encontrado.")

# Exemplo de uso:
encontrar_headers_duplicados("riboswitch_100.fa")
```

```python
from collections import defaultdict

def encontrar_sequencias_duplicadas(fasta_path):
    sequencias = defaultdict(list)
    header = None
    seq = ''

    with open(fasta_path, 'r') as f:
        for line in f:
            line = line.strip()
            if line.startswith('>'):
                if header and seq:
                    sequencias[seq].append(header)
                header = line
                seq = ''
            else:
                seq += line
        # Última sequência
        if header and seq:
            sequencias[seq].append(header)

    # Verificar duplicatas
    duplicadas = {seq: headers for seq, headers in sequencias.items() if len(headers) > 1}

    if duplicadas:
        print("Sequências duplicadas encontradas:\n")
        for i, (seq, headers) in enumerate(duplicadas.items(), 1):
            print(f"🔁 Duplicata {i}:")
            print(f"  Ocorre {len(headers)} vezes.")
            print("  Headers:")
            for h in headers:
                print(f"    {h}")
            print(f"  Sequência: {seq[:60]}... [tamanho: {len(seq)}]\n")
    else:
        print("✅ Nenhuma sequência duplicada encontrada.")

# Exemplo de uso:
encontrar_sequencias_duplicadas("riboswitch_100.fa")
```

```python
import pickle

def parse_fasta(fasta_path):
    query_to_seq = {}
    with open(fasta_path, 'r') as f:
        current_query = None
        current_seq = []

        for line in f:
            line = line.strip()
            if not line:
                continue  # ignora linhas vazias
            if line.startswith('>'):
                if current_query and current_seq:
                    full_seq = ''.join(current_seq).strip()
                    query_to_seq[current_query] = full_seq
                current_query = line[1:]  # remove '>'
                current_seq = []
            else:
                current_seq.append(line)

        # última sequência
        if current_query and current_seq:
            full_seq = ''.join(current_seq).strip()
            query_to_seq[current_query] = full_seq

    return query_to_seq

def parse_out_file(out_path, query_to_seq):
    data_dict = {}
    with open(out_path, 'r') as f:
        lines = [line.strip() for line in f if line.strip()]  # remove vazios

    i = 0
    while i < len(lines):
        if lines[i].startswith('>'):
            query = lines[i][1:]
            seq = query_to_seq.get(query)

            estrutura = None
            energia = None

            # procurar a próxima linha com parênteses (estrutura)
            j = i + 1
            while j < len(lines):
                line = lines[j]
                if any(x in line for x in '(){}[].'):
                    partes = line.split()
                    if len(partes) >= 2 and '(' in partes[0]:
                        estrutura = partes[0]
                        try:
                            energia = float(partes[-1].strip('(){}[]<>'))  # remove símbolos
                        except ValueError:
                            energia = None
                        break
                j += 1

            if seq and estrutura and energia is not None:
                data_dict[seq] = [estrutura, energia]
        i += 1

    faltantes = set(query_to_seq.values()) - set(data_dict.keys())
    print(f"⚠️  Sequências sem estrutura/energia: {len(faltantes)}")

    return data_dict


def main():
    fasta_file = "riboswitch_100.fa"
    out_file = "riboswitch_100.out"
    output_pkl = "riboswitch_100.pkl"

    query_to_seq = parse_fasta(fasta_file)
    data_dict = parse_out_file(out_file, query_to_seq)

    with open(output_pkl, 'wb') as f:
        pickle.dump(data_dict, f)

    print(f"\n✅ Dicionário salvo em '{output_pkl}' com {len(data_dict)} entradas.")
    print(f"📦 Total de sequências no FASTA: {len(query_to_seq)}")

if __name__ == "__main__":
    main()
```
