O algoritmo utiliza duas camadas de janelas deslizantes para explorar diferentes contextos de uma sequência genômica. Primeiro, aplica-se uma janela externa fixa (por exemplo, 200 nt com passo de 20 nt) ao longo de toda a sequência longa; essa janela define um contexto local maior. Em seguida, dentro de cada janela externa, são geradas várias janelas internas menores, cujos tamanhos variam progressivamente (por exemplo, de 95 até 200 nt, de 5 em 5) e que também se deslocam com um passo menor (por exemplo, 5 nt). Para cada janela interna calcula-se a sobreposição com o riboswitch conhecido, a porcentagem de cobertura e um rótulo (riboswitch ou background). Dessa forma, o método captura simultaneamente variação de posição e de escala, permitindo analisar o mesmo trecho genômico em múltiplos comprimentos e deslocamentos.

Exemplo: em uma sequência de 545 nt, a janela externa de 200 nt com passo 20 gera 18 contextos diferentes. Dentro da primeira janela externa (posições 0–200), o algoritmo cria janelas internas de tamanhos 95, 100, 105 … até 200 nt, cada uma deslizando de 5 em 5 nucleotídeos. Uma janela interna de 95 nt pode começar em 0, 5, 10 … até 105, totalizando 22 posições possíveis; já uma janela interna de 200 nt tem apenas 1 posição (0). Cada uma dessas sub-janelas é então avaliada quanto à presença e porcentagem de cobertura do riboswitch.

```python
import csv

# ===== PARÂMETROS =====
genome_fasta = "1_riboswitch/test_riboswitches_E_coli_200up_200down_GCN.fa"
ribo_fasta   = "1_riboswitch/test_riboswitches_E_coli.fa"
out_fasta    = "1_riboswitch/riboswitch_windows_dual.fa"
out_csv      = "1_riboswitch/riboswitch_windows_dual.csv"

#genome_fasta = "2_riboswitch/test_riboswitches_E_coli_200up_200down_GCN.fa"
#ribo_fasta   = "2_riboswitch/test_riboswitches_E_coli.fa"
#out_fasta    = "2_riboswitch/riboswitch_windows_dual.fa"
#out_csv      = "2_riboswitch/riboswitch_windows_dual.csv"

#genome_fasta = "3_riboswitch/test_riboswitches_E_coli_200up_200down_GCN.fa"
#ribo_fasta   = "3_riboswitch/test_riboswitches_E_coli.fa"
#out_fasta    = "3_riboswitch/riboswitch_windows_dual.fa"
#out_csv      = "3_riboswitch/riboswitch_windows_dual.csv"


#genome_fasta = "4_riboswitch/test_riboswitches_E_coli_200up_200down_GCN.fa"
#ribo_fasta   = "4_riboswitch/test_riboswitches_E_coli.fa"
#out_fasta    = "4_riboswitch/riboswitch_windows_dual.fa"
#out_csv      = "4_riboswitch/riboswitch_windows_dual.csv"

#genome_fasta = "5_riboswitch/test_riboswitches_E_coli_200up_200down_GCN.fa"
#ribo_fasta   = "5_riboswitch/test_riboswitches_E_coli.fa"
#out_fasta    = "5_riboswitch/riboswitch_windows_dual.fa"
#out_csv      = "5_riboswitch/riboswitch_windows_dual.csv"

#genome_fasta = "6_riboswitch/test_riboswitches_E_coli_200up_200down_GCN.fa"
#ribo_fasta   = "6_riboswitch/test_riboswitches_E_coli.fa"
#out_fasta    = "6_riboswitch/riboswitch_windows_dual.fa"
#out_csv      = "6_riboswitch/riboswitch_windows_dual.csv"

#genome_fasta = "7_riboswitch/test_riboswitches_E_coli_200up_200down_GCN.fa"
#ribo_fasta   = "7_riboswitch/test_riboswitches_E_coli.fa"
#out_fasta    = "7_riboswitch/riboswitch_windows_dual.fa"
#out_csv      = "7_riboswitch/riboswitch_windows_dual.csv"

#genome_fasta = "8_riboswitch/test_riboswitches_E_coli_200up_200down_GCN.fa"
#ribo_fasta   = "8_riboswitch/test_riboswitches_E_coli.fa"
#out_fasta    = "8_riboswitch/riboswitch_windows_dual.fa"
#out_csv      = "8_riboswitch/riboswitch_windows_dual.csv"

# Janela EXTERNA
EXT_WINDOW_SIZE = 200
EXT_STEP_SIZE   = 20

# Janelas INTERNAS (dentro da externa)
INT_MIN_SIZE = 95
INT_MAX_SIZE = 200
INT_SIZE_STEP = 5
INT_STEP_SIZE = 5

THRESHOLD   = 75.0


def read_fasta(path):
    seqs = {}
    header = None
    chunks = []
    with open(path, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            if line.startswith(">"):
                if header is not None:
                    seqs[header] = "".join(chunks).upper()
                header = line[1:]
                chunks = []
            else:
                chunks.append(line)
        if header is not None:
            seqs[header] = "".join(chunks).upper()
    return seqs


def revcomp(seq):
    comp = str.maketrans("ACGTUacgtuNn", "TGCAAtgcaaNn")
    return seq.translate(comp)[::-1]


def find_riboswitch_coords(long_seq, ribo_seq):
    start = long_seq.find(ribo_seq)
    if start != -1:
        return start, start + len(ribo_seq), "+"

    rc = revcomp(ribo_seq)
    start = long_seq.find(rc)
    if start != -1:
        return start, start + len(ribo_seq), "-"

    return None


def overlap_len(a_start, a_end, b_start, b_end):
    left = max(a_start, b_start)
    right = min(a_end, b_end)
    return max(0, right - left)


def write_fasta_record(out, header, seq, width=80):
    out.write(f">{header}\n")
    for i in range(0, len(seq), width):
        out.write(seq[i:i+width] + "\n")


def ribo_position_inside_window(w_start, w_end, ribo_start, ribo_end):
    """
    Posição (1-based) do trecho do riboswitch que cai dentro da janela.
    Retorna (0,0) se não houver overlap.
    """
    ov_start = max(w_start, ribo_start)
    ov_end   = min(w_end, ribo_end)
    if ov_end <= ov_start:
        return 0, 0
    pos_ini = (ov_start - w_start) + 1
    pos_fim = (ov_end - w_start)          # ov_end exclusivo
    return pos_ini, pos_fim


def ribo_position_inside_riboswitch(w_start, w_end, ribo_start, ribo_end):
    """
    Posição (1-based) do trecho do riboswitch ORIGINAL que cai dentro da janela.
    Retorna (0,0) se não houver overlap.
    """
    ov_start = max(w_start, ribo_start)
    ov_end   = min(w_end, ribo_end)
    if ov_end <= ov_start:
        return 0, 0
    ribo_pos_ini = (ov_start - ribo_start) + 1
    ribo_pos_fim = (ov_end - ribo_start)       # ov_end exclusivo
    return ribo_pos_ini, ribo_pos_fim


def iter_internal_windows(ext_start, ext_end,
                          int_min=95, int_max=200, size_step=5, int_step=5):
    """
    Gera (w_start, w_end, w_size, offset_in_ext) para todas as janelas internas
    contidas na janela externa [ext_start, ext_end).
    """
    ext_len = ext_end - ext_start
    max_size_allowed = min(int_max, ext_len)

    size = int_min
    while size <= max_size_allowed:
        # offset dentro da externa: 0..(ext_len-size), passo int_step
        for off in range(0, ext_len - size + 1, int_step):
            w_start = ext_start + off
            w_end = w_start + size
            yield w_start, w_end, size, off
        size += size_step


def varredura_dupla(
    genome_seqs,
    ribo_seqs,
    out_fasta,
    out_csv,
    ext_window_size=200,
    ext_step_size=20,
    int_min_size=95,
    int_max_size=200,
    int_size_step=5,
    int_step_size=5,
    threshold=70.0
):
    seq_id = 1
    cont = 1

    with open(out_fasta, "w", encoding="utf-8") as out, \
         open(out_csv, "w", encoding="utf-8", newline="") as csv_out:

        writer = csv.writer(csv_out)
        writer.writerow([
            "window_id",
            "seq_id",
            "sequence_name",
            "strand",
            "outside_window",
            "inside_window",
            "window_start",
            "window_end",
            "riboswitch_start",
            "riboswitch_end",
            "overlap_nt",
            "riboswitch_len",
            "pct_identificacao",
            "position_in_window",
            "label"
        ])


        for header, long_seq in genome_seqs.items():

            if header not in ribo_seqs:
                print(f"[ERRO] Header não encontrado no ribo_fasta: {header}")
                continue

            ribo_seq = ribo_seqs[header]
            coords = find_riboswitch_coords(long_seq, ribo_seq)

            if coords is None:
                print(f"[AVISO] Riboswitch não encontrado em: {header}")
                continue

            ribo_start, ribo_end, strand = coords
            ribo_len = ribo_end - ribo_start

            total_ext = 0
            total_int = 0
            ext_index = 1

            # ===== janelas EXTERNAS =====
            for ext_start in range(0, len(long_seq) - ext_window_size + 1, ext_step_size):
                ext_end = ext_start + ext_window_size
                total_ext += 1

                # ===== janelas INTERNAS (dentro da externa) =====
                for w_start, w_end, w_size, off_in_ext in iter_internal_windows(
                    ext_start, ext_end,
                    int_min=int_min_size,
                    int_max=int_max_size,
                    size_step=int_size_step,
                    int_step=int_step_size
                ):
                    ov = overlap_len(w_start, w_end, ribo_start, ribo_end)
                    pct = 100.0 * ov / ribo_len if ribo_len > 0 else 0.0
                    pct_int = int(round(pct))

                    win_ini, win_fim = ribo_position_inside_window(w_start, w_end, ribo_start, ribo_end)
                    win_pos_str = f"{win_ini}-{win_fim}" if win_ini != 0 else "0-0"

                    label = "riboswitch" if pct >= threshold else "genomicbackground"

                    # Header FASTA: inclui info da janela externa + interna
                    # - ext: intervalo genômico e índice da externa
                    # - int: tamanho e offset dentro da externa
                    new_header = (
                        f"{cont}_{seq_id}_{header}_"
                        f"ext_{ext_start}-{ext_end}_int_{w_size}_off_{off_in_ext}_"
                        f"percentage_{pct_int}_position_{win_pos_str} {label}"
                    )

                    window_seq = long_seq[w_start:w_end]
                    write_fasta_record(out, new_header, window_seq)

                    writer.writerow([
                        cont,
                        seq_id,
                        header,
                        strand,
                        ext_index,     # outside_window
                        w_size,        # inside_window (tamanho)
                        w_start,
                        w_end,
                        ribo_start,
                        ribo_end,
                        ov,
                        ribo_len,
                        round(pct, 2),
                        f"position_{win_pos_str}",
                        label
                    ])
                    
                    cont += 1
                    total_int += 1
                ext_index += 1

            print(
                f"[OK] {header} | strand={strand} | riboswitch={ribo_start}-{ribo_end} | "
                f"ext_windows={total_ext} | int_windows={total_int}"
            )
            seq_id += 1


# ===== EXECUÇÃO =====
genome_seqs = read_fasta(genome_fasta)
ribo_seqs   = read_fasta(ribo_fasta)

varredura_dupla(
    genome_seqs,
    ribo_seqs,
    out_fasta=out_fasta,
    out_csv=out_csv,
    ext_window_size=EXT_WINDOW_SIZE,
    ext_step_size=EXT_STEP_SIZE,
    int_min_size=INT_MIN_SIZE,
    int_max_size=INT_MAX_SIZE,
    int_size_step=INT_SIZE_STEP,
    int_step_size=INT_STEP_SIZE,
    threshold=THRESHOLD
)

print(f"[SAÍDA] FASTA: {out_fasta}")
print(f"[SAÍDA] CSV : {out_csv}")
```

```python
import pickle

def parse_fasta(fasta_path):
    query_to_seq = {}
    with open(fasta_path, 'r') as f:
        current_query = None
        current_seq = []

        for line in f:
            line = line.strip()
            if not line:
                continue  # ignora linhas vazias
            if line.startswith('>'):
                if current_query and current_seq:
                    full_seq = ''.join(current_seq).strip()
                    query_to_seq[current_query] = full_seq
                current_query = line[1:]  # remove '>'
                current_seq = []
            else:
                current_seq.append(line)

        # última sequência
        if current_query and current_seq:
            full_seq = ''.join(current_seq).strip()
            query_to_seq[current_query] = full_seq

    return query_to_seq

def parse_out_file(out_path, query_to_seq):
    data_dict = {}
    with open(out_path, 'r') as f:
        lines = [line.strip() for line in f if line.strip()]  # remove vazios

    i = 0
    while i < len(lines):
        if lines[i].startswith('>'):
            query = lines[i][1:]
            seq = query_to_seq.get(query)

            estrutura = None
            energia = None

            # procurar a próxima linha com parênteses (estrutura)
            j = i + 1
            while j < len(lines):
                line = lines[j]
                if any(x in line for x in '(){}[].'):
                    partes = line.split()
                    if len(partes) >= 2 and '(' in partes[0]:
                        estrutura = partes[0]
                        try:
                            energia = float(partes[-1].strip('(){}[]<>'))  # remove símbolos
                        except ValueError:
                            energia = None
                        break
                j += 1

            if seq and estrutura and energia is not None:
                data_dict[seq] = [estrutura, energia]
        i += 1

    faltantes = set(query_to_seq.values()) - set(data_dict.keys())
    print(f"Sequências sem estrutura/energia: {len(faltantes)}")

    return data_dict


def main():
    fasta_file = "1_riboswitch/riboswitch_windows_dual.fa"
    out_file = "1_riboswitch/riboswitch_windows_dual.out"
    output_pkl = "1_riboswitch/riboswitch_windows_dual.pkl"

    #fasta_file = "2_riboswitch/riboswitch_windows_dual.fa"
    #out_file = "2_riboswitch/riboswitch_windows_dual.out"
    #output_pkl = "2_riboswitch/riboswitch_windows_dual.pkl"
    
    #fasta_file = "3_riboswitch/riboswitch_windows_dual.fa"
    #out_file = "3_riboswitch/riboswitch_windows_dual.out"
    #output_pkl = "3_riboswitch/riboswitch_windows_dual.pkl"
    
    #fasta_file = "4_riboswitch/riboswitch_windows_dual.fa"
    #out_file = "4_riboswitch/riboswitch_windows_dual.out"
    #output_pkl = "4_riboswitch/riboswitch_windows_dual.pkl"
    
    #fasta_file = "5_riboswitch/riboswitch_windows_dual.fa"
    #out_file = "5_riboswitch/riboswitch_windows_dual.out"
    #output_pkl = "5_riboswitch/riboswitch_windows_dual.pkl"
    
    #fasta_file = "6_riboswitch/riboswitch_windows_dual.fa"
    #out_file = "6_riboswitch/riboswitch_windows_dual.out"
    #output_pkl = "6_riboswitch/riboswitch_windows_dual.pkl"
    
    #fasta_file = "7_riboswitch/riboswitch_windows_dual.fa"
    #out_file = "7_riboswitch/riboswitch_windows_dual.out"
    #output_pkl = "7_riboswitch/riboswitch_windows_dual.pkl"
       
    #fasta_file = "8_riboswitch/riboswitch_windows_dual.fa"
    #out_file = "8_riboswitch/riboswitch_windows_dual.out"
    #output_pkl = "8_riboswitch/riboswitch_windows_dual.pkl"
    
    query_to_seq = parse_fasta(fasta_file)
    data_dict = parse_out_file(out_file, query_to_seq)

    with open(output_pkl, 'wb') as f:
        pickle.dump(data_dict, f)

    print(f"\nDicionário salvo em '{output_pkl}' com {len(data_dict)} entradas.")
    print(f"Total de sequências no FASTA: {len(query_to_seq)}")

if __name__ == "__main__":
    main()

```

```python
from Bio import SeqIO
from collections import Counter, defaultdict

def diagnosticar_duplicatas_fasta(caminho_fasta, modo="id", mostrar_exemplos=30):
    """
    Diagnostica duplicatas em um arquivo FASTA.

    Parâmetros:
    - caminho_fasta (str): caminho do arquivo .fa/.fasta
    - modo (str): "id" para duplicação por ID (rec.id)
                 "seq" para duplicação por sequência (str(rec.seq))
    - mostrar_exemplos (int): quantos exemplos imprimir
    
    Retorna:
    - dict com estatísticas
    """
    if modo not in {"id", "seq"}:
        raise ValueError("modo deve ser 'id' ou 'seq'")

    records = list(SeqIO.parse(caminho_fasta, "fasta"))

    if modo == "id":
        chaves = [rec.id for rec in records]
        titulo = "ID"
    else:
        chaves = [str(rec.seq) for rec in records]
        titulo = "SEQUÊNCIA"

    total = len(chaves)
    contagem = Counter(chaves)

    # pega apenas duplicados
    duplicados = {k: v for k, v in contagem.items() if v > 1}

    # distribuição (2x, 3x, 4x...)
    hist = Counter(duplicados.values())

    print("=" * 80)
    print(f"Diagnóstico de duplicatas no FASTA: {caminho_fasta}")
    print(f"Modo: {modo} (por {titulo})")
    print("=" * 80)
    print(f"Total de sequências no arquivo: {total}")
    print(f"{titulo}s únicos: {len(contagem)}")
    print(f"{titulo}s duplicados (>1): {len(duplicados)}")

    if duplicados:
        print("\nDistribuição de duplicação:")
        for vezes in sorted(hist.keys()):
            print(f" - {hist[vezes]} {titulo}(s) aparecem {vezes}x")

        print(f"\nExemplos de {titulo}s duplicados (até {mostrar_exemplos}):")
        for k, v in sorted(duplicados.items(), key=lambda x: (-x[1], x[0]))[:mostrar_exemplos]:
            print(f" - {k}  ->  {v}x")
    else:
        print("\nNenhuma duplicata encontrada!")

    return {
        "total": total,
        "unicos": len(contagem),
        "n_duplicados": len(duplicados),
        "duplicados": duplicados,
        "histograma": dict(hist),
    }

stats_seq = diagnosticar_duplicatas_fasta("3_riboswitch/riboswitch_windows_dual.fa", modo="id")
stats_seq = diagnosticar_duplicatas_fasta("3_riboswitch/riboswitch_windows_dual.fa", modo="seq")

#stats_seq = diagnosticar_duplicatas_fasta("train/set2/train_SET2.fa", modo="id")

#stats_seq = diagnosticar_duplicatas_fasta("train/set3/train_SET3.fa", modo="id")
```
